### Step 3: Train YOLOv8 on GPU
Fine-tune the YOLOv8 model for 50 epochs with early stopping patience of 15.

```python
from ultralytics import YOLO

# Initialize YOLOv8 nano model
model = YOLO('yolov8n.pt')

# Train model
results = model.train(
    data='pothole_data/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    project='/content/drive/MyDrive/pothole',
    name='run1',
    patience=15
)
```

### Step 4: Validate and Compute Metrics

```python
metrics = model.val()
print("=== Quantitative Validation Metrics ===")
print(f"mAP50     : {metrics.box.map50:.4f}")
print(f"mAP50-95  : {metrics.box.map:.4f}")
print(f"Precision : {metrics.box.mp:.4f}")
print(f"Recall    : {metrics.box.mr:.4f}")
```
